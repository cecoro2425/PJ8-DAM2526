## Carga y preparación del conjunto de datos

En esta sección se cargan los datos desde un archivo CSV y se realiza una primera preparación de la información.  
Se incluyen transformaciones básicas como la conversión de tipos de datos y la normalización de eventos climáticos extremos, con el objetivo de facilitar su análisis posterior.


```python
from analisis import (
    ProyectoAgricola,
    AnalisisClimatico,
    AnalisisSocioeconomico
)

# Setup: carregar el CSV i preparar columnes útils
import pandas as pd
import numpy as np

# Llegeix el CSV. Nota: alguns camps són strings numèrics; els convertirem a numèrics.
#df = pd.read_csv('climate_change.csv')
df = pd.read_csv('climate_change_refact.csv')

exWeatherEvt = df['Extreme_Weather_Events'].astype(float)
event_min, event_max = exWeatherEvt.min(), exWeatherEvt.max()

# Evitem divisió per zero: si rc_max == rc_min, posem normalized a 0.0
if event_max == event_min:
    df['Extreme_Weather_Events_Normalize'] = 0.0
else:
    df['Extreme_Weather_Events_Normalize'] = (exWeatherEvt - event_min) / (event_max - event_min)

print('Files, Columnes:', df.shape)
print('Columnes:', list(df.columns))
df.head(5)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('climate_change_refact.csv')
def filtrar_agricultura(
    df,
    country,
    region,
    crop,
    years,
):
    base_columns = [
        'Year',
        'Country',
        'Region',
        'Crop_Type',
        'Average_Temperature_C',
        'Adaptation_Strategies',
        'Economic_Impact_Million_USD',
        'Fertilizer_Use_KG_per_HA',
        'Pesticide_Use_KG_per_HA',
        'Irrigation_Access_%',
        'Crop_Yield_MT_per_HA'
    ]

    mask = (
        df['Year'].isin(years) &
        (df['Country'] == country) &
        (df['Region'] == region) &
        (df['Crop_Type'] == crop)
    )

    resultado = (
        df.loc[mask, base_columns]
          .sort_values(by='Year')
          .reset_index(drop=True)
    )

    return resultado
```

```python
def graficar_serie_temporal(
    df,
    y_col,
    titulo,
    y_label
):
    plt.figure(figsize=(10,5))
    
    plt.plot(
        df['Year'],
        df[y_col],
        marker='o'
    )
    
    plt.title(titulo)
    plt.xlabel('Year')
    plt.ylabel(y_label)
    plt.grid(True)
    plt.show()
```

## Análisis de relaciones entre variables productivas y económicas

En esta sección se estudian relaciones entre diferentes factores agrícolas, como el acceso al riego, el uso de fertilizantes y las estrategias de adaptación, y su impacto en el rendimiento y los resultados económicos.


```python
years = [1990, 1995, 2000, 2005, 2011, 2015, 2020, 2024]

datos = filtrar_agricultura(
    df=df,
    country='Brazil',
    region='South',
    crop='Corn',
    years=years,
)

graficar_serie_temporal(
    df=datos,
    y_col='Crop_Yield_MT_per_HA',
    titulo='Fertilizantes en trigo – Sur de Brasil',
    y_label='kg por hectárea'
)

#datos_trigo_sur_brasil.head(20)
```

```python
plt.figure(figsize=(6,5))
plt.scatter(
    datos['Irrigation_Access_%'],
    datos['Economic_Impact_Million_USD']
)
plt.xlabel('Acceso a riego (%)')
plt.ylabel('Impacto económico (M USD)')
plt.title('Riego vs impacto económico – Trigo en Brasil')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.scatter(
    datos['Fertilizer_Use_KG_per_HA'],
    datos['Crop_Yield_MT_per_HA']
)
plt.xlabel('Fertilizantes (kg/ha)')
plt.ylabel('Rendimiento (MT/ha)')
plt.title('Fertilizantes vs rendimiento – Trigo en Brasil')
plt.grid(True)
plt.show()
```

## Conclusiones

A partir del análisis realizado, se observa que, aunque el cambio climático influye indirectamente en la utilización de recursos agrícolas y en los costes de producción, los impactos económicos más significativos suelen estar asociados a factores globales.

Entre estos factores destacan los desastres naturales, las fluctuaciones del valor de la moneda, la escasez de recursos derivada de conflictos políticos y los eventos climáticos extremos, como variaciones anómalas en las precipitaciones.

Ejemplos recientes, como los episodios de lluvias intensas ocurridas durante la borrasca Leonardo, evidencian cómo estos fenómenos pueden alterar la planificación agrícola y la gestión de recursos, amplificando los efectos económicos.

En conclusión, el impacto socioeconómico del cambio climático en la agricultura no actúa de forma aislada, sino que se ve condicionado por el contexto global, económico y geopolítico.
